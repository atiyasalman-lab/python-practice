```python
mat3 = np.random.randint(10, size=(4,4))
mat4 = np.random.randint(10, size=(4,3))
```

```python
np.dot(mat3,mat4)
```

```python
# Take the latest two arrays, perform a transpose on one of them and then perform 
# a dot product on them both
np.dot(mat3.T,mat4)
```

```python
# Compare the two arrays with '>='
mat5 >= mat6
```
