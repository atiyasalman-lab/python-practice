```python
import pandas as pd
a = pd.Series([2,4,6,8,10])
b = pd.Series([1,3,5,7,9])
ds= a + b
print("Add Two Series")
print(ds)
print("Subtract two Series")
ds = a - b 
print(ds)
print("multiply two series")
ds = a*b
print(ds)
print("divide two series")
ds = a/b
print(ds)
```

```python
import pandas as pd
ds1 = pd.Series([2, 4, 6, 8, 10])
ds2 = pd.Series([1, 3, 5, 7, 10])
print("Series1:")
print(ds1)
print("Series2:")
print(ds2)
print("Compare the elements of the said Series:")
print("Equals:")
print(ds1 == ds2)
print("Greater than:")
print(ds1 > ds2)
print("Less than:")
print(ds1 < ds2)
```
