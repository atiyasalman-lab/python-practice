```python
hr_sal = df[df["dept"] == "HR"]["salary"]
hr_sal
```

```python
fin_salary = df[df["dept"] == "Finance"]["salary"]
fin_salary
```

```python
# bivariate = numerical - categorical
plt.boxplot([hr_sal , it_sal, fin_salary], tick_labels = ["HR", "IT", "Finance"])
plt.show()
```

```python
salary_by_dept = df.groupby("dept")["salary"].sum()
salary_by_dept
```

```python
df["experience"] = [1,3,5,6,9,8,6,5,9,3,12,1.2,5,10,11,5,2,1,6,6]
```

```python
df.head()
```

```python
# bubble plot
plt.scatter(df["Age"], df["salary"], s = df["experience"]*50, color = "skyblue", edgecolors = "black")
plt.show()
```

```python
plt.scatter(df["Age"], df["salary"], s = df["experience"]*50, color = "skyblue", edgecolors = "black")
plt.title("Age vs Salary vs Experience")
plt.xlabel("Age")
plt.ylabel("salary")
plt.show()
```

```python
plt.scatter(df["Age"], df["salary"], c = df["dept"].map({"HR":"yellow", "IT":"blue", "Finance": "orange"}))
plt.xlabel("Age")
plt.ylabel("salary")
plt.title("Age vs salary vs Dept")
plt.legend()
plt.show()
```

```python
#object oriented API --- allows subplots and multiple plots in single figure
```

```python
fig, axs = plt.subplots(1,3, figsize = (10,5))

#line plot
axs[0].plot(df["Age"], df["salary"], color = "red", marker = "+", linewidth = "2", markersize = "2" )
axs[0].grid()
axs[0].set_title("Line plot")
axs[0].set_xlabel("Age")
axs[0].set_ylabel("salary")

#histogram
axs[1].hist(df["salary"], bins= 5 , color = "green" )
axs[1].set_title("Histogram")
axs[1].set_xlabel("salary")
axs[1].set_ylabel("ylabel")

#box plot
axs[2].boxplot(df["salary"])
axs[2].set_title("Boxxxx pLott")
axs[2].set_xlabel("salary")

plt.show
plt.savefig("multipleplots.png")   #if i also want to save the multiplot figure
```
